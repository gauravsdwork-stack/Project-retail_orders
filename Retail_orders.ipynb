```python
#importing csv file
import pandas as pd
df = pd.read_csv(r"C:\Users\LENOVO\kaggle-retail-project\orders.csv", na_values=['Not Available', 'unknown'])
df.head()
```

```python
#changing columns to lower cases
df.columns = df.columns.str.lower().str.replace(" ", "_")
df.head()
```

```python
#creating a new column named discount
df['discount']=df['list_price']*df['discount_percent']/100
```

```python
df
```

```python
#creating a new column named sale price 
df['sale_price']=df['list_price']-df['discount']
df.head()
```

```python
#creating a new column named profit
df['profit']=df['sale_price']-df['cost_price']
df.head()
```

```python
df.dtypes
```

```python
df['order_date']=pd.to_datetime(df['order_date'], format="%Y-%m-%d")
```

```python
#drop cost price, list price and discont percent columns
df.drop(columns=['cost_price', 'list_price', 'discount_percent'], inplace=True)
```
